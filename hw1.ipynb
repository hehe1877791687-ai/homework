A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:
   由于题目给出的数组满足 A[1] > A[2]且 A[n-1] < A[n]，这意味着数组在最开始是下降的，在最后是上升的。则必然存在至少一个局部最小值。我们可以使用二分查找的思想，我们每次选取中间点mid并观察其与右侧邻居的关系：如果A[mid]>A[mid+1]，说明当前位置是下降趋势，局部最小值一定在mid的右侧（包括 mid+1）。如果A[mid] < A[mid+1]说明从mid到 mid+1是上升趋势，这意味着mid可能是局部最小值，或者局部最小值在mid的左侧。通过这种方式，我们每次将搜索空间减半，从而达到O(log n)的时间复杂度。

````python
# add your code here
def minimas(nums):
    n=len(nums)
    if n==0:return[]
    if n==1:return[nums[0]]
    res=[]
    if nums[0]<nums[1]:res.append(nums[0])
    for i in range(1,n-1):
        if nums[i-1]>nums[i]<nums[i+1]:res.append(nums[i])
    if nums[-1]<nums[-2]:res.append(nums[-1])
    return res

a=[9,3,7,2,1,4,5]
print(f"局部最小值:{minimas(a)}")

# your algorithm time complexity is:O(log n)
````

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:
我们可以使用一个栈来模拟星球的运行状态。遍历数组中的每一个星球：处理向右运动的星球（正数）：因为它们永远不会撞上此时在它们左侧的星球，直接将其压入栈中。
处理向左运动的星球（负数）：当遇到向左的星球时，它可能会与栈顶的向右星球发生连续碰撞。我们需要一个 while 循环进行检查：如果栈为空，或者栈顶是向左运动的星球（负数），说明不会发生碰撞，直接将该负数压入栈中。如果栈顶是向右运动的星球（正数），则发生碰撞：比较两者质量（绝对值）：若栈顶质量 < 当前星球质量：栈顶被消灭（出栈），继续对比下一个栈顶。若栈顶质量 == 当前星球质量：当前向左运动的星球消失。停止检查。若栈顶质量 > 当前星球质量：当前向左运动的星球消失。停止检查。时间复杂度为O（n）。

````python
# add your code here
def p(ps):
    s=[]
    for p in ps:
        a=1
        while s and p<0 and s[-1]>0:
            if s[-1]<-p:
                s.pop()
                continue
            elif s[-1]==-p:
                a=0
                break
            else:
                a=0
                break
        if a:s.append(p)
    return s

t1=[-3,-6,2,8,5,-8,9,-2,1]
print(f"输入:{t1}")
print(f"输出:{p(t1)}")

t=[23,-8,9,-3,-7,9,-23,22]
print(f"输入:{t}")
print(f"输出:{p(t)}")

# your algorithm time complexity is:时间复杂度为O（n）
````

````python
# add your code here
def ks(c, its):
    l=len(its)
    dp=[[0]*(c+1)for _ in range(l+1)]
    for i in range(1,l+1):
        w,v=its[i-1]
        for j in range(c+1):
            dp[i][j]=dp[i-1][j]if w>j else max(dp[i-1][j],dp[i-1][j-w]+v)
    return dp[l][c]

its=[(5,10),(4,40),(6,30),(3,50)]
m=9
r=ks(m,its)
print(f"最大价值:{r}")
# your algorithm time complexity is:
````

输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:
二叉搜索树的中序遍历的结果必然是一个严格递增的序列。当有两个节点的值被错误交换后，这个递增序列中就会出现“异常”。异常情况：遍历时，如果发现当前节点的值小于前一个节点的值,这即是一个“逆序对”。寻找交换节点：如果只交换了相邻的两个节点，只会发现一个逆序对。如果交换了不相邻的节点，会发现两个逆序对。我们只需要记录这些逆序对中涉及的节点，进行一次交换即可恢复.由于我们需要找出被错误交换的两个节点，我们必须检查整棵树的所有节点,算法的时间复杂度为O(N)，其中N是二叉搜索树中节点的总数。

````python
# add your code here
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def recover(root):
    pre = TreeNode(float('-inf'))
    a = None
    b = None

    def dfs(n):
        nonlocal pre,a,b
        if not n:return
        dfs(n.left)
        if n.val<pre.val:
            if not a:
                a=pre
                b=n
            else:
                b=n
        pre=n
        dfs(n.right)

    dfs(root)
    a.val,b.val = b.val,a.val

def build(arr):
    if not arr:return None
    root=TreeNode(arr[0])
    q=[root]
    i=1
    while q and i<len(arr):
        node=q.pop(0)
        if arr[i]!=None:
            node.left=TreeNode(arr[i])
            q.append(node.left)
        i+=1
        if i<len(arr) and arr[i]!=None:
            node.right=TreeNode(arr[i])
            q.append(node.right)
        i+=1
    return root

def show(root):
    res=[]
    q=[root]
    while q:
        node=q.pop(0)
        if node:
            res.append(node.val)
            q.append(node.left)
            q.append(node.right)
        else:
            res.append(None)
    while res and res[-1]==None:
        res.pop()
    print(res)

root1=build([1,3,None,None,2])
recover(root1)
show(root1)

root2=build([3,1,4,None,None,2])
recover(root2)
show(root2)
    # # your algorithm time complexity is:时间复杂度为O（n)
````

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:
---
1.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-1.jpg"> </div>

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:                 该题借助了大模型

该问题的核心在于如何将树的路径转换为数字。我们可以观察到，从根节点到叶子节点的路径构建过程本质上是一个深度优先搜索的过程。逻辑推导：状态传递：在遍历树的过程中，我们记录从根节点到当前节点的路径所代表的数字（设为 current_sum）。状态更新：当从父节点移动到子节点时，新的路径数字计算公式为：currentsum_next = current_sum_prev *10 + node.val例如：从节点 1 到 2，之前的值是 1，则 1 * 10 + 2 = 12。递归终止条件：如果当前节点为空，返回 0。如果当前节点为叶子节点（即 left 和 right 均为 None），直接返回当前累计的 current_sum。递归合并：将左右子树返回的结果相加，即得到最终和.

时间复杂度：O(N)
其中N是二叉树中节点的总数，算法需要对每个节点进行一次访问。

````python
class TreeNode:
    def __init__(self,val=0,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right

def s(root):
    def d(n,c):
        if not n:return 0
        c=c*10+n.val
        if not n.left and not n.right:return c
        return d(n.left,c)+d(n.right,c)
    return d(root,0)

r1=TreeNode(1,TreeNode(2),TreeNode(3))
print(f"示例1:{s(r1)}")

r2=TreeNode(4,TreeNode(9,TreeNode(5),TreeNode(1)),TreeNode(0))
print(f"示例2:{s(r2)}")
# your algorithm time complexity is:
````
