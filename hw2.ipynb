## B 长跑

```python
# add your code here

#include <stdio.h>
#include <stdlib.h>
#include <string.h>

#define INF 0x3f3f3f3f

typedef struct {
    int pos;
    int cost;
} Station;

int cmp(const void *a, const void *b) {
    Station *s1 = (Station *)a;
    Station *s2 = (Station *)b;
    if (s1->pos != s2->pos) {
        return s1->pos - s2->pos;
    }
    return s1->cost - s2->cost;
}

Station stations[2005];
Station unique_stations[2005];
int dp[2005];

int main() {
    int N, L, Maxn, S;
    
    while (scanf("%d %d %d %d", &N, &L, &Maxn, &S) == 4) {
        for (int i = 0; i < N; i++) {
            scanf("%d %d", &stations[i].pos, &stations[i].cost);
        }
        
        qsort(stations, N, sizeof(Station), cmp);
        
        int u_cnt = 0;
        for (int i = 0; i < N; i++) {
            if (stations[i].pos > L || stations[i].pos < 0) continue;
            
            if (u_cnt > 0 && unique_stations[u_cnt - 1].pos == stations[i].pos) {
                if (stations[i].cost < unique_stations[u_cnt - 1].cost) {
                    unique_stations[u_cnt - 1].cost = stations[i].cost;
                }
            } else {
                unique_stations[u_cnt] = stations[i];
                u_cnt++;
            }
        }
        
        for (int i = u_cnt; i > 0; i--) {
            unique_stations[i] = unique_stations[i - 1];
        }
        unique_stations[0].pos = 0;
        unique_stations[0].cost = 0;
        u_cnt++;
        
        for (int i = 0; i < u_cnt; i++) {
            dp[i] = INF;
        }
        dp[0] = 0;
        
        for (int i = 1; i < u_cnt; i++) {
            for (int j = 0; j < i; j++) {
                if (unique_stations[i].pos - unique_stations[j].pos <= Maxn) {
                    if (dp[j] != INF && dp[j] + unique_stations[i].cost < dp[i]) {
                        dp[i] = dp[j] + unique_stations[i].cost;
                    }
                }
            }
        }
        
        int min_total_cost = INF;
        for (int i = 0; i < u_cnt; i++) {
            if (L - unique_stations[i].pos <= Maxn) {
                if (dp[i] < min_total_cost) {
                    min_total_cost = dp[i];
                }
            }
        }
        
        if (min_total_cost <= S) {
            printf("Yes\n");
        } else {
            printf("No\n");
        }
    }
    
    return 0;
}
```

## C 最长回文

```python
## add your code here

#include<bits/stdc++.h>
#define int long long
#define ull unsigned long long
using namespace std;
const int N = 3e5 + 10;
const int MOD = 999998639; 
const int P = 13331;
string a , b;
int pa[N] , pb[N] , res = 1;
ull pre[N] , suf[N] , power[N];
string Manacher(string a , int *p)
{
    string t = "$#";
    for(auto i : a) t += i , t += '#';
    int mx = 0 , id = 0 ;
    int len = t.size() , ans = 0; 
    for(int i = 1 ; i < len ; i ++)
    {
        p[i] = mx > i ? min(p[2 * id - i] , mx - i) : 1;
        while(t[i + p[i]] == t[i - p[i]]) p[i] ++ ;
        if(mx < i + p[i]) mx = i + p[i] , id = i;
        ans = max(ans , p[i] - 1);
    }
    res = max(res , ans);
    return t;
}
ull get_hash1(int l , int r)
{
    if(l > r) return -999;
    return (pre[r] - pre[l - 1] * power[r - l + 1] % MOD + MOD) % MOD;
}
ull get_hash2(int l , int r)
{
    if(l > r) return -888;
    return (suf[l] - suf[r + 1] * power[r - l + 1] % MOD + MOD) % MOD;
}
void init(int n)
{
    power[0] = 1;
    for(int i = 1 ; i < N - 5 ; i ++) power[i] = power[i - 1] * P % MOD;
    pre[0] = a[0] ;
    for(int i = 1 ; i < n ; i ++) pre[i] = (pre[i - 1] * P + a[i]) % MOD ;
    suf[n - 1] = b[n - 1];
    for(int i = n - 2 ; i >= 0 ; i --) suf[i] = (suf[i + 1] * P + b[i]) % MOD; 
}
signed main()
{
    int n ;
    cin >> n >> a >> b;
    a = Manacher(a , pa) , b = Manacher(b , pb);
    n = n * 2 + 2;
    init(n);
    int ans = 1;
    for(int i = 2 ; i <= n ; i ++)
    {
        int add = 0;
        int len = max(pa[i] , pb[i - 2]);
        int sa = i - len + 1  , sb = i - 2 + len - 1;
        int l = 0 , r = min(sa - 1 , n - sb);
        while(l <= r)
        {
            int mid = l + r >> 1;
            int l1 = max(0LL , sa - mid) , r1 = min(n , sa - 1);
            int l2 = max(sb + 1 , 0LL) , r2 = min(n , sb + mid);
            if(get_hash1(l1 , r1) == get_hash2(l2 , r2)) 
            l = mid + 1 , add = mid;
            else r = mid - 1; 
        }
        ans = max(ans , add + len - 1);
    }
    cout << ans << '\n';
    return 0;
}
```

## D 优惠券

```python
## add your code here
#include <set>
#include <queue>
#include <vector>
#include <cstdio>
#include <cstring>
#include <algorithm>
using namespace std;

const int MAX_SIZE = 1000005;

set<int> available_pos;
int in_time[MAX_SIZE];
int out_time[MAX_SIZE];

int main()
{
    int op_count, idx, val, first_error;
    char op_type;
    
    while (scanf("%d", &op_count) != EOF)
    {
        memset(in_time, 0, sizeof(in_time));
        memset(out_time, 0, sizeof(out_time));
        
        while (!available_pos.empty())
            available_pos.clear();
        
        first_error = -1;
        
        for (idx = 1; idx <= op_count; idx++)
        {
            scanf(" %c", &op_type);
            
            if (op_type == '?')
            {
                available_pos.insert(idx);
                continue;
            }
            
            scanf("%d", &val);
            
            if (op_type == 'I')
            {
                if (in_time[val])
                {
                    set<int>::iterator it = available_pos.upper_bound(in_time[val]);
                    if (available_pos.empty() || it == available_pos.end())
                    {
                        if (first_error == -1)
                            first_error = idx;
                    }
                    else
                        available_pos.erase(it);
                }
                in_time[val] = idx;
            }
            
            if (op_type == 'O')
            {
                if (in_time[val] == 0)
                {
                    set<int>::iterator it = available_pos.upper_bound(out_time[val]);
                    if (available_pos.empty() || it == available_pos.end())
                    {
                        if (first_error == -1)
                            first_error = idx;
                    }
                    else
                        available_pos.erase(it);
                }
                in_time[val] = 0;
                out_time[val] = idx;
            }
        }
        printf("%d\n", first_error);
    }
    return 0;
}
```

## E 任意点

```python
## add your code here
#include <stdio.h>
#include <stdlib.h>

#define MAXN 105

int parent[MAXN];
int x[MAXN], y[MAXN];

void init(int n) {
    for (int i = 1; i <= n; i++) {
        parent[i] = i;
    }
}

int find(int i) {
    if (parent[i] == i)
        return i;
    return parent[i] = find(parent[i]);
}

void union_sets(int i, int j) {
    int root_i = find(i);
    int root_j = find(j);
    if (root_i != root_j) {
        parent[root_i] = root_j;
    }
}

int main() {
    int n;
    if (scanf("%d", &n) != 1) return 0;

    init(n);

    for (int i = 1; i <= n; i++) {
        scanf("%d %d", &x[i], &y[i]);
    }

    for (int i = 1; i <= n; i++) {
        for (int j = i + 1; j <= n; j++) {
            if (x[i] == x[j] || y[i] == y[j]) {
                union_sets(i, j);
            }
        }
    }

    int components = 0;
    for (int i = 1; i <= n; i++) {
        if (parent[i] == i) {
            components++;
        }
    }

    if (n == 0) {
        printf("0\n");
    } else {
        printf("%d\n", components - 1);
    }

    return 0;
}
```

## I 直方图最大矩形

```python
## add your code here
import java.util.*;

public class Solution {
    public int largestRectangleArea (int[] heights) {
        int n=heights.length;
        int r=0;
        for(int i=0;i<n;i++)
        {
            int s=-1;
            for(int j=i-1;j>=0;j--)
            {
                if(heights[j]<heights[i])
                {
                    s=j;
                    break;
                }
            }
            int e=n;
            for(int j=i+1;j<n;j++)
            {
                if(heights[j]<heights[i])
                {
                    e=j;
                    break;
                }
            }
            r=Math.max(r,(e-s-1)*heights[i]);
        }
        return r;
    }
}
```

## J 消防局的设立

```python
## add your code here
#include <stdio.h>
#include <stdlib.h>
#define MAXN 1005

int fa[MAXN],dep[MAXN],vis[MAXN];
int head[MAXN],to[MAXN*2],nxt[MAXN*2],tot;
int n,ans;

int cmp(const void*a,const void*b){
    return dep[*(int*)b]-dep[*(int*)a];
}

void add(int u,int v){
    to[++tot]=v;
    nxt[tot]=head[u];
    head[u]=tot;
}

void dfs(int u,int f){
    fa[u]=f;
    dep[u]=dep[f]+1;
    for(int i=head[u];i;i=nxt[i]){
        int v=to[i];
        if(v!=f)dfs(v,u);
    }
}

void mark(int u,int f,int d){
    vis[u]=1;
    if(d==2)return;
    for(int i=head[u];i;i=nxt[i]){
        int v=to[i];
        if(v!=f)mark(v,u,d+1);
    }
}

int main(){
    scanf("%d",&n);
    tot=0;ans=0;
    for(int i=2;i<=n;i++){
        int x;
        scanf("%d",&x);
        add(i,x);
        add(x,i);
    }
    dfs(1,0);
    int id[MAXN];
    for(int i=1;i<=n;i++)id[i]=i;
    qsort(id+1,n,sizeof(int),cmp);
    for(int i=1;i<=n;i++){
        int u=id[i];
        if(!vis[u]){
            int g=fa[fa[u]];
            if(!g)g=1;
            mark(g,0,0);
            ans++;
        }
    }
    printf("%d\n",ans);
    return 0;
}
```
